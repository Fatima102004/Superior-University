```python
CAP_X = 3
CAP_Y = 4
GOAL = 2

def next_states(x, y):
    moves = []
    if x < CAP_X:
        moves.append((CAP_X, y, "Fill Jug X"))
    if y < CAP_Y:
        moves.append((x, CAP_Y, "Fill Jug Y"))
    if x > 0:
        moves.append((0, y, "Empty Jug X"))
    if y > 0:
        moves.append((x, 0, "Empty Jug Y"))
    if x > 0 and y < CAP_Y:
        shift = min(x, CAP_Y - y)
        moves.append((x - shift, y + shift, f"Pour {shift} from X → Y"))
    if y > 0 and x < CAP_X:
        shift = min(y, CAP_X - x)
        moves.append((x + shift, y - shift, f"Pour {shift} from Y → X"))
    return moves

def dfs_water_jug(start=(0, 0)):
    stack = [(start, [])]
    seen = set()
    while stack:
        (x, y), path = stack.pop()
        if x == GOAL or y == GOAL:
            return path + [((x, y), "Goal Reached")]
        if (x, y) in seen:
            continue
        seen.add((x, y))
        for nx, ny, action in next_states(x, y):
            stack.append(((nx, ny), path + [((nx, ny), action)]))
    return None

path = dfs_water_jug()

print("=== Water Jug DFS Solution ===\n")
if path:
    for i in range(len(path)):
        state, action = path[i]
        print(f"Move {i+1}: {action} | Resulting state: {state}")
    print(f"\n Success! One jug now has exactly {GOAL} liters.")
else:
    print("No valid sequence of moves was found.")
```
