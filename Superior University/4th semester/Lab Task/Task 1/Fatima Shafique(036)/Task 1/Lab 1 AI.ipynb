```python
def fillNaMean(df, categorical_cols):
    for col in categorical_cols:
        df[col] = df[col].fillna(df[col].mode()[0])

categorical_columns = ['Age','Cabin','Embarked']
fillNaMean(df, categorical_columns)
```

```python
def changeFloattoInt64(cols):
    for i in cols:
       df[i] = df[i].astype('int64')

columns = ['Fare', 'Age']
changeFloattoInt64(columns)
```

```python
def encodeCols(cols):
    for i in cols:
        temp = pd.DataFrame({i:df[i].unique()})
        data_LE = LabelEncoder()
        data_LE.fit(np.ravel(temp))
        df[i] = data_LE.transform(df[i])

columns = ['Embarked', 'Cabin', 'Ticket','Sex','Name']
encodeCols(columns)
```

```python
df.info()
```

```python
df.drop('PassengerId', axis=1, inplace=True)
df.drop('Name', axis=1, inplace=True)
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)
```
