```python
import cv2

photo = cv2.imread(r"E:\Dagestani_man_and_woman.jpg")


gray_img = cv2.cvtColor(photo, cv2.COLOR_BGR2GRAY)

blur_img = cv2.GaussianBlur(photo, (7,7), 0)

edge_img = cv2.Canny(photo, 100, 200)

small_img = cv2.resize(photo, (300, 300))

flip_img = cv2.flip(photo, 1)

rot_img = cv2.rotate(photo, cv2.ROTATE_90_CLOCKWISE)

cv2.imwrite("gray_image.jpg", gray_img)
cv2.imwrite("blur_image.jpg", blur_img)
cv2.imwrite("edge_image.jpg", edge_img)
cv2.imwrite("small_image.jpg", small_img)
cv2.imwrite("flip_image.jpg", flip_img)
cv2.imwrite("rotate_image.jpg", rot_img)
```
